```python

import ast

def safe_convert(text):

    # If already a list, return it directly
    if isinstance(text, list):
        return text

    # If missing, return an empty list
    if pd.isna(text):
        return []

    # Convert string into Python list
    return ast.literal_eval(text)
```

```python

import ast
import pandas as pd

def parse_data(value):

    if isinstance(value, list):
        return value

    if pd.isna(value):
        return []

    if isinstance(value, str):
        try:
            return ast.literal_eval(value)
        except (ValueError, SyntaxError):
            return []

    return []


def extract_names(value):

    data = parse_data(value)

    names = []

    for item in data:

        if isinstance(item, dict):
            if 'name' in item:
                names.append(item['name'])

        elif isinstance(item, str):
            names.append(item)

    return names


print("Safe extraction functions created!")
```

```python

movies['genres'] = movies['genres'].apply(extract_names)

movies['keywords'] = movies['keywords'].apply(extract_names)

display(movies[['title', 'genres', 'keywords']].head())
```

```python

def extract_cast(value):

    data = parse_data(value)

    actors = []

    for item in data:

        if isinstance(item, dict) and 'name' in item:
            actors.append(item['name'])

        elif isinstance(item, str):
            actors.append(item)

    return actors[:3]


movies['cast'] = movies['cast'].apply(extract_cast)

display(movies[['title', 'cast']].head())
```

```python

def extract_director(value):

    data = parse_data(value)

    for item in data:

        if isinstance(item, dict):
            if item.get('job') == 'Director':
                return [item['name']]

    return []


director_column = 'crew' if 'crew' in movies.columns else 'director'

movies['director'] = movies[director_column].apply(extract_director)

display(movies[['title', 'director']].head())
```

```python
movies['tags'] = (
    movies['overview']
    + movies['genres']
    + movies['keywords']
    + movies['cast']
    + movies['director']
)

display(movies[['title', 'tags']].head())
```

```python
movies['tags'] = movies['tags'].apply(
    lambda x: " ".join(x.split())
)

print("Text preprocessing completed!")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

print("TF-IDF imported successfully!")
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words='english'
)

tfidf_matrix = tfidf.fit_transform(movies['tags'])

print("TF-IDF transformation completed!")
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
print("Number of movies:", tfidf_matrix.shape[0])
print("Number of features:", tfidf_matrix.shape[1])

print("\nFirst movie vector:")
print(tfidf_matrix[0])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

print("Cosine Similarity imported successfully!")
```

```python
similarity = cosine_similarity(tfidf_matrix)

print("Similarity matrix created successfully!")
print("Matrix shape:", similarity.shape)
```

```python
print("Selected Movie:", movies['title'].iloc[0])

similarity_scores = list(enumerate(similarity[0]))

similarity_scores = sorted(
    similarity_scores,
    key=lambda x: x[1],
    reverse=True
)

print("\nTop 10 similar movie indices:")

for item in similarity_scores[1:11]:
    print(item)
```

```python
def recommend_movies(movie_name, top_n=10):

    # Find the selected movie
    movie_name = movie_name.lower()

    matches = movies[
        movies['title'].str.lower() == movie_name
    ]

    if matches.empty:
        print("Movie not found. Please check the movie title.")
        return

    # Get movie index
    movie_index = matches.index[0]

    # Get similarity scores
    distances = similarity[movie_index]

    # Sort movies by similarity
    movie_list = sorted(
        list(enumerate(distances)),
        key=lambda x: x[1],
        reverse=True
    )

    print(f"\nMovies similar to {movies.iloc[movie_index]['title']}:\n")

    count = 0

    for index, score in movie_list:

        # Exclude selected movie
        if index == movie_index:
            continue

        print(
            f"{count + 1}. {movies.iloc[index]['title']} "
            f"(Similarity: {score:.3f})"
        )

        count += 1

        if count == top_n:
            break
```

```python

def recommend_movies(movie_name, top_n=10):

    # Find matching movie
    matches = movies[
        movies['title'].str.lower() == movie_name.lower()
    ]

    if matches.empty:
        print("Movie not found!")
        return pd.DataFrame()

    movie_index = matches.index[0]

    # Calculate similarity scores
    distances = similarity[movie_index]

    # Sort by similarity
    movie_indices = sorted(
        enumerate(distances),
        key=lambda x: x[1],
        reverse=True
    )

    recommendations = []

    for index, score in movie_indices:

        if index == movie_index:
            continue

        recommendations.append({
            'Movie': movies.iloc[index]['title'],
            'Similarity Score': round(float(score), 4)
        })

        if len(recommendations) == top_n:
            break

    return pd.DataFrame(recommendations)
```
